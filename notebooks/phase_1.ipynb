# Stock Market Prediction System - Phase 1: Problem Definition & Dataset Exploration
**Academic College ML Project SOP**

### 1. Problem Statement
Financial markets are inherently non-stationary and volatile. The objective of this project is to develop a machine learning pipeline capable of predicting:
1. **Tomorrow's High Price** (Regression)
2. **Tomorrow's Low Price** (Regression)
3. **Tomorrow's Expected Price Range** (High - Low)
4. **Market Direction** (Binary Classification: Bullish vs. Bearish)
5. **Trading Signal** (Multiclass Classification: BUY, HOLD, SELL)

> **Disclaimer**: *This project is developed solely for educational and academic demonstration. Stock market predictions are non-deterministic and do not constitute financial advice.*

```python
import os
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set visualization styles
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (12, 6)

symbol = 'TCS.NS'
print(f'Fetching historical data for {symbol}...')
df = yf.download(symbol, period='5y', interval='1d', progress=False)

# Flatten MultiIndex columns if present in newer yfinance
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

print('Data download complete. Shape:', df.shape)
df.head()
```

### 2. Dataset Exploration & Structural Analysis
Inspecting the data types, dimensions, summary statistics, and identifying missing data.

```python
print('--- DATASET INFORMATION ---')
print(df.info())

print('\n--- DATA TYPES ---')
print(df.dtypes)

print('\n--- MISSING VALUE SUMMARY ---')
print(df.isnull().sum())

print('\n--- DESCRIPTIVE STATISTICS ---')
df.describe()
```

### 3. Historical Price and Volume Visualization

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8), sharex=True, gridspec_kw={'height_ratios': [3, 1]})

ax1.plot(df.index, df['Close'], label='Close Price', color='#2563eb', linewidth=1.5)
ax1.set_title(f'{symbol} Historical Close Price (5-Year Trend)', fontsize=14, fontweight='bold')
ax1.set_ylabel('Price (INR ₹)', fontsize=12)
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.bar(df.index, df['Volume'], color='#64748b', alpha=0.7, width=1.0)
ax2.set_title('Trading Volume', fontsize=12)
ax2.set_xlabel('Date', fontsize=12)
ax2.set_ylabel('Volume', fontsize=12)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
